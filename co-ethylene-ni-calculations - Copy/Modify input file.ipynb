```python
import os
import re
from collections import OrderedDict

directory = r"C:\Users\George\Desktop\Research_UCLA\CIC\Computational Work\Project cluster\structures\co-ethylene-ni-calculations"

LANL2DZ_METALS = {'Ni'}

HEADER_TEMPLATE = """%mem=16000MB
%nprocshared=16
%chk={base_name}.chk
#p opt freq genecp pop=nbo m06

{title}

0 1
"""

def build_basis_block(elements):
    lanl_metals = sorted(e for e in elements if e in LANL2DZ_METALS)
    non_metals = sorted(e for e in elements if e not in LANL2DZ_METALS)

    basis = ""
    if non_metals:
        basis += f"\n{' '.join(non_metals)} 0\ndef2svpp\n****\n"
    for metal in lanl_metals:
        basis += f"{metal} 0\nLANL2DZ\n****\n\n{metal} 0\nLANL2DZ\n"
    return basis.strip()

def extract_coordinates_and_elements(lines):
    coord_lines = []
    elements = []
    for line in lines:
        if re.match(r"^\s*[A-Z][a-z]?\s+-?\d+\.\d+\s+-?\d+\.\d+\s+-?\d+\.\d+", line):
            coord_lines.append(line.strip())
            elements.append(line.strip().split()[0])
        elif coord_lines:
            break  # stop at first non-coordinate line after coordinates
    return coord_lines, list(OrderedDict.fromkeys(elements))

for filename in os.listdir(directory):
    if filename.endswith(".com") and "_formatted" not in filename:
        filepath = os.path.join(directory, filename)
        with open(filepath, 'r') as f:
            lines = f.readlines()

        coords, elements = extract_coordinates_and_elements(lines)

        if 'Pd' in elements:
            print(f"❌ Skipped due to Pd: {filename}")
            continue

        base_name = os.path.splitext(filename)[0]
        title = base_name.replace("-", " ")

        header = HEADER_TEMPLATE.format(base_name=base_name, title=title)
        basis_block = build_basis_block(elements)

        new_content = header + "\n".join(coords) + "\n\n" + basis_block + "\n\n\n"

        new_filename = os.path.join(directory, f"{base_name}_formatted.com")
        with open(new_filename, 'w') as f:
            f.write(new_content)

        print(f"✅ Cleaned and written: {new_filename}")

print("🎉 All files processed.")
```
